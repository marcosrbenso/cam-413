```r
# Criar um vetor
v <- c(4,5,123,52,53,4,2,5)
v
```

```r
# Criar um data.frame
df <- data.frame(
  v1 = v,
  v2 = runif(8)
)

df
```

```r
# Selecionar uma coluna do data
df$v1
```

```r
# plotar com ggplot2
ggplot(df,aes(v1,v2))+geom_line()
```
